```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'NOME': ['Ana', 'Bruno', 'Carlos'],
    'IDADE': [25, 30, 35]
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'NOME': ['Diana', 'Eduardo', 'Fernanda'],
    'IDADE': [28, 32, 40]
})
df2
```

```python
df_notas = pd.DataFrame({
    'MATEMÁTICA': [8.5, 7.0, 9.0],
    'PORTUGUÊS':  [7.5, 8.0, 6.5]
})
df_notas
```

```python
resultado_vertical = pd.concat([df1, df2])
resultado_vertical
```

```python
resultado_horizontal = pd.concat([df1, df_notas], axis=1)
resultado_horizontal
```
