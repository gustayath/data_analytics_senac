```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'CATEGORIA': ['A', 'B', 'A', 'B', 'A'],
    'VENDAS': [100, 200, 150, 300, 120],
    'QUANTIDADE': [1, 2, 1, 3, 2]
})
df
```

```python
# Agrupa por categoria e faz a média

df.groupby('CATEGORIA')['VENDAS'].describe()
```

```python
# Agrupa por categoria e faz a média

df.groupby('CATEGORIA')['QUANTIDADE'].describe()
```

```python
# Soma por categoria (apenas colunas numéricas)

df.groupby('CATEGORIA').sum()
```

```python
# Agrupa por múltiplas colunas

df.groupby(['CATEGORIA', 'QUANTIDADE'], as_index=False)['VENDAS'].sum()
```
