```python
import pandas as pd
import mysql.connector
from mysql.connector import Error
```

```python
try:
    # Estabelece a conexão
    conexao = mysql.connector.connect(
        host = 'localhost',
        user = 'root',
        password = '',
        database = 'escola'
    )

    if conexao.is_connected():
        print("Conectado com sucesso!")

        # Cria um cursor (objeto que permite executar comandos)
        cursor = conexao.cursor()

        # Executa um consulta
        cursor.execute("SELECT * FROM alunos LIMIT 5")

        # Busca os resultados 
        resultados = cursor.fetchall()
        for i in resultados:
            print(i)

except Error as e:
    print(f"Erro ao conectar ou executar: {e}")

finally:
    # Fecha o cursor (se foi criado)
    if 'cursor' in locals():
        try:
            cursor.close()
        except Exception:
            pass # Se já estiver fechado ou ocorrer erro, ignore

    # Fecha a conexão (se foi criada e ainda está conectada)
    if 'conexao' in locals() and conexao.is_connected():
        conexao.close()
        print("conexão encerrada.")
```
