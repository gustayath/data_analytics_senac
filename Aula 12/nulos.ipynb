```python
import pandas as pd
```

```python
# DataSet com valores nulos

df = pd.read_csv(r"C:\Users\gustavo.yath\Desktop\DADOS\funcionarios - Null.csv")
df
```

```python
# Visualização de valores nulos

df.isna().sum()
```

```python
# Remove valores nulos da série 'departamento'
# dropna remove a linha por inteiro

df_removido = df.dropna(subset=['departamento', 'lucro'])
df_removido
```

```python
# Visualização de valores nulos

df_removido.isna().sum()
```

```python
# Sobrescreve todos os valores vazios por 'Nenhum' e '0'
# fillna sobrescreve valores

df_preenchido = df
df_preenchido = df_preenchido.fillna({
    'departamento': "Nenhum",
    'lucro': 0
})
df_preenchido
```

```python
# Visualização de valores nulos

df_preenchido.isna().sum()
```

```python
# Para preencher apenas os valores da coluna 'Departamento'

df_departamento_preenchido = df
df_departamento_preenchido['departamento'] = df_departamento_preenchido['departamento'].fillna("Nenhum")
df_departamento_preenchido
```

```python
# Visualização de valores nulos

df_departamento_preenchido.isna().sum()
```
